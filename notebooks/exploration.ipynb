```python
df["temperature_rolling"] = df["temperature_c"].rolling(window=20).mean()
```

```python
print(df['temperature_rolling'])
```

```python
df["temperature_std"] = df["temperature_c"].rolling(window=20).std()
```

```python
print(df["temperature_std"])
```

```python
df[["temperature_c", "temperature_rolling", "temperature_std"]].head(50)
```

```python
plt.figure(figsize=(12,5))
plt.plot(df["temperature_std"])
plt.title("Temperature Instability")
plt.xlabel("Time")
plt.ylabel("Standard Deviation")
plt.show()
```

```python
plt.figure(figsize=(12,5))
plt.plot(df["temperature_rolling"], label = "Temperature Trend")
plt.plot(df["temperature_std"], label = "Instability")
plt.legend()
plt.show()
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df["trend_scaled"] = scaler.fit_transform(
    df[["temperature_rolling"]]
)

df["instability_scaled"] = scaler.fit_transform(
    df[["temperature_std"]]
)

plt.figure(figsize=(12,5))

plt.plot(df["trend_scaled"], label="Trend")
plt.plot(df["instability_scaled"], label="Instability")

plt.legend()
plt.show()
```
